
# Implementation of Snake and Ladder Using BFS

## Aim
To implement the Snake and Ladder game using Breadth-First Search (BFS) and find the minimum number of dice throws required to reach the final square.

## Objectives

1. To understand the Breadth-First Search (BFS) algorithm.
2. To represent the Snake and Ladder board as a graph.
3. To use a queue to explore possible moves level by level.
4. To calculate the minimum number of dice throws required to reach square 100.
5. To understand how BFS finds the shortest path in an unweighted graph.

## Theory

Snake and Ladder is a board game in which a player starts at square 1 and tries to reach square 100 by rolling a dice.

- **Breadth-First Search (BFS):** BFS explores possible moves level by level.
- **Queue:** A queue follows FIFO (First In, First Out) and stores the squares to be explored.
- **Graph:** Each square represents a node, and each possible dice throw represents an edge.
- **Snakes:** If a player lands on a snake's head, the player moves down to its tail.
- **Ladders:** If a player lands at the bottom of a ladder, the player moves up to its top.

BFS explores paths in increasing order of dice throws. Therefore, it can find the minimum number of throws required to reach square 100.


```python

from collections import deque

# Snakes and ladders: starting square -> destination square
snakes = {
    17: 7,
    54: 34,
    62: 19,
    64: 60,
    87: 24,
    93: 73,
    95: 75,
    99: 78
}

ladders = {
    4: 14,
    9: 31,
    20: 38,
    28: 84,
    40: 59,
    51: 67,
    63: 81,
    71: 91
}

def snake_and_ladder_bfs():
    start = 1
    destination = 100

    # Queue stores: (current square, number of dice throws)
    queue = deque([(start, 0)])

    # Track visited squares to avoid repeated exploration
    visited = {start}

    # Store the previous square for reconstructing the path
    parent = {start: None}

    while queue:
        position, throws = queue.popleft()

        # Destination reached
        if position == destination:
            path = []
            current = destination

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()
            return throws, path

        # Try every possible dice result from 1 to 6
        for dice in range(1, 7):
            next_position = position + dice

            # Do not move beyond square 100
            if next_position > destination:
                continue

            # Apply a snake or ladder, if present
            if next_position in snakes:
                next_position = snakes[next_position]
            elif next_position in ladders:
                next_position = ladders[next_position]

            # Add each square to the queue only once
            if next_position not in visited:
                visited.add(next_position)
                parent[next_position] = position
                queue.append((next_position, throws + 1))

    return -1, []


# Run the BFS algorithm
minimum_throws, path = snake_and_ladder_bfs()

print("SNAKE AND LADDER USING BFS")
print("-" * 35)
print("Starting square:", 1)
print("Destination square:", 100)
print("Minimum dice throws:", minimum_throws)
print("One shortest path:", " -> ".join(map(str, path)))
```

```python

print("SNAKES")
for head, tail in sorted(snakes.items()):
    print(f"{head} -> {tail}")

print("\nLADDERS")
for bottom, top in sorted(ladders.items()):
    print(f"{bottom} -> {top}")
```

## Algorithm

1. Start the program.
2. Define the snake and ladder positions.
3. Initialize a queue with square 1 and zero dice throws.
4. Mark square 1 as visited.
5. Remove the first square from the queue.
6. If the square is 100, return the number of throws and the path.
7. Otherwise, try all dice values from 1 to 6.
8. Ignore moves beyond square 100.
9. Apply the snake or ladder transition when necessary.
10. Add each unvisited destination square to the queue.
11. Repeat until square 100 is reached or the queue becomes empty.
12. Display the minimum throws and path.
13. Stop.



## Time Complexity

- **Time Complexity:** O(N × D), where N is the number of board squares and D is the number of possible dice outcomes (6).
- **Space Complexity:** O(N), for the queue, visited set, and parent dictionary.

For a standard board of 100 squares, both are effectively linear in the board size because the dice has only six possible outcomes.
